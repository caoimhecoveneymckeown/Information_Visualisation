#### 3. Rename columns

```python
for df in [ghg, air]:
    df.columns = ["statistic", "year", "sector", "unit", "value"]
    df["code"] = df["statistic"].str.extract(r"\(([^)]*)\)$")
```

```python
print(ghg["code"].unique())
print(air["code"].unique())
```

#### 4. Only count sectors once

```python
totals = ["Total NACE and Household activities", "Total NACE activities", "Total Household activities"]
ghg = ghg[~ghg["sector"].isin(totals)]
air = air[~air["sector"].isin(totals)]
```

```python
gas_names = {"CO2": "CO2", "CH4_CO2E": "CH4", "N2O_CO2E": "N2O",
             "HFC_CO2E": "HFCs", "PFC_CO2E": "PFCs", "NF3_SF6_CO2E": "SF6 + NF3"}
pollutant_names = {"NH3": "NH3", "CO": "CO", "NOX": "NOx", "NMV0C": "NMVOC",
                   "PM10": "PM10", "PM2_5": "PM2.5", "SOX_SO2E": "SOx"}
```

```python
sectors = pd.DataFrame({"sector": ghg["sector"].unique()})
sectors["sector_name"] = sectors["sector"].str.replace(r"\s*\([^)]*\)$", "", regex=True)
```

#### 5. Create a sector table

```python
sectors = pd.DataFrame({"sector": ghg["sector"].unique()})
sectors["sector_name"] = sectors["sector"].str.replace(r"\s*\([^)]*\)$", "", regex=True)
```

```python
sectors["nace"] = sectors["sector"].str.extract(r"\((\d+)[^)]*\)$")[0].astype(float)

conditions = [
    sectors["nace"].isna(),                                    # household rows have no code
    sectors["nace"] <= 3,
    sectors["nace"].between(35, 39),
    (sectors["nace"] <= 9) | sectors["nace"].between(41, 43),
    sectors["nace"].between(10, 33),
    sectors["nace"].between(49, 53),
]
groups = ["Households", "Agriculture, forestry & fishing", "Electricity, water & waste",
          "Mining & construction", "Manufacturing", "Transport & storage"]
sectors["group"] = np.select(conditions, groups, default="Services")
```

#### 6. Combine GHG totals and air pollutants into one table

```python
ghg_total = ghg[ghg["code"] == "GHG"].copy()
ghg_total["measure"] = "GHG"

air_kt = air.copy()
air_kt["measure"] = air_kt["code"].map(pollutant_names)
air_kt["value"] = air_kt["value"] / 1000        

keep = ["year", "sector", "measure", "value"]
emissions = pd.concat([ghg_total[keep], air_kt[keep]])
emissions = emissions.merge(sectors, on="sector")
```

#### 7. Share percentage of national total, per year and measure for each sector

```python
yearly_total = emissions.groupby(["year", "measure"])["value"].transform("sum")
emissions["share_pct"] = 100 * emissions["value"] / yearly_total
```

#### 8. Index relative to 2009

```python
base = emissions[emissions["year"] == 2009][["sector", "measure", "value"]]
base = base.rename(columns={"value": "value_2009"})
emissions = emissions.merge(base, on=["sector", "measure"])
emissions["index_2009"] = 100 * emissions["value"] / emissions["value_2009"]
emissions.loc[emissions["value_2009"] == 0, "index_2009"] = np.nan 
```

#### 9. GHG split by gas type

```python
gas_mix = ghg[ghg["code"].isin(gas_names)].copy()
gas_mix["gas"] = gas_mix["code"].map(gas_names)
gas_mix = gas_mix.merge(sectors, on="sector")
gas_mix = gas_mix[["year", "sector_name", "group", "gas", "value"]]
gas_mix = gas_mix.rename(columns={"value": "kt_co2e"})
```

#### 10. Sector shares in the latest year, one column per measure

```python
latest = emissions[emissions["year"] == emissions["year"].max()]
shares_latest = latest.pivot_table(index=["sector_name", "group"],
                                   columns="measure", values="share_pct").reset_index()
```
